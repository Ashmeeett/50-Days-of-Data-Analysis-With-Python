
1. Using pandas, import the website_data_analysis.cvs dataset and find the average number of visits per website.

```python

import pandas as pd

df = pd.read_csv("Datasets_50_Days_Of_Data_Analysis/website_data_analysis.csv")
df.head()
```

```python

df_average_visit_per_site = df.groupby('website')['visits'].mean()
df_average_visit_per_site.head()
```


2. Add another column to the DataFrame that calculates the average number of website views per unique visitor.

```python
df["views_per_unique_visit"] = df["visits"] / df["unique_visitors"]
df.head()
```


3. Using Matplotlib, create a bar graph to visualize the top 5 websites with the highest number of page views.

```python

import matplotlib.pyplot as plt

# Sort the website by visits in descending order
top_5_websites = df.sort_values(by="visits", ascending=False).head(5)

plt.figure(figsize=(12, 8))
plt.bar(top_5_websites["website"], top_5_websites["visits"])
plt.xlabel("Website")
plt.ylabel("Visits")
plt.title("Top 5 Websites by Visits", fontsize=20)
plt.show()
```


4. Using pandas, group the data by the day of the week and find the average bounce rate for each day.

```python

average_bounce_rate_by_day = df.groupby("days_of_week")["bounce_rate"].mean()
average_bounce_rate_by_day
```


5. Using seaborn, create a line plot to show the trend of unique visitors over time.

```python

import seaborn as sns

# Group data by days of the week
df_group = df.groupby('days_of_week')['unique_visitors'].mean()

# Plot the data
plt.figure(figsize = (12, 8))
sns.lineplot(x = df_group.index, y = df_group.values)
plt.xlabel("Day of Week")
plt.ylabel("Unique Visitors")
plt.title("Unique Visitors Over Time", fontsize = 20)
plt.show()
```


6. Using pandas, group the data by "day_of_week" and "referral_source" columns and find the average number of visits and revenue for each group.

```python

grouped_df = df.groupby(['days_of_week','referral_source'])[['visits','revenue']].mean()
grouped_df.head()
```


7. Using pandas, calculate the revenue rate for each referral source and create a pie chart to visualize the breakdown of revenue rate by referral source. Which referral source brought it the most revenue?

```python
df["revenue_rate%"]= (df["revenue"]/(sum(df["revenue"])))*100

group_data = df.groupby("referral_source")["revenue_rate%"].sum()

# Plotting with pandas and matplotlib
plt.figure(figsize=(12, 8))
group_data.plot.pie(autopct='%.1f%%')
plt.title("Revenue Rate by Referral Source", fontsize = 20)
plt.show()
```
