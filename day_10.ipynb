
2. Using NumPy, write a code that returns all the numbers in the array that are over 25.


3. Here is some data below:

hours_worked = [ 8, 8, 10, 11, 7]

```python
days = ["mon", "tue",  "wed",  "thu",  "Fri"]
hours_worked = [ 8, 8, 10, 11, 7]

# Creating arrays
days = np.array(days)
hours_worked = np.array(hours_worked)

# Using Boolean to get the 8 hour days.
days_8hrs = days[hours_worked == 8]
days_8hrs
```


4. Write another code (using Boolean indexing) to return to return the day of the week with the highest number of hours worked.

```python
max_hours_index = np.argmax(hours_worked)
print(f'Day with the max hours worked is, {days[max_hours_index]}')
```

5. Write another code to return all the days of the week, where over 9 hours of work was achieved.

```python

days_over_9hrs = days[hours_worked > 9]
days_over_9hrs
```
