6. Extract the year from the "X1 transaction date" column and 
create a year column called "X1 Transaction Year." Drop the 
"X1 transaction date" from the DataFrame. The "X1 
Transaction Year" should be the only date column. 

```python

def convert_column(column):
    column =  column.astype("str")
    # Split the column into two columns
    df_copy [["X1 Transaction Year", "Date"]] = column.str.split(".", expand = True).astype(int)
    df_copy.drop(columns=["X1 transaction date","Date" ], inplace=True)
    return df_copy.head()

convert_column(df_copy["X1 transaction date"])
```


7. Shuffling a DataFrame is important in machine learning to avoid any bias that may be introduced when the data is ordered in a certain way. Our data seem to have a particular order. The prices and sizes are ordered from smallest to largest. This may create bias. Using scikit-learn, shuffle the DataFrame.

```python
df_shuffled = shuffle(df_copy, random_state = 42)
df_shuffled.head()
```


8. Write a code to separate the DataFrame into X and y variables. X for feature columns and y for the target column. Use scikit-learn's train_test_split function to split the X and y variables into training and test sets. Make the test set to be 10% of the data.

```python
X = df_shuffled.drop(columns=["Y house price of unit area"])
y = df_shuffled["Y house price of unit area"]

X_train, X_test, y_train, y_test = train_test_split(X, y, 
                                                    test_size=0.1, 
                                                    random_state=23)
```


9. Using scikit-learn LinearRegression, select the model for your data based on the size of your dataset. Ensure that your data is scaled before fitting. Use the r2_score, and MSE(mean squared error) metrics to evaluate the model. What conclusion can you draw from the score results?

```python

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)


reg = LinearRegression()
reg.fit(X_train, y_train)


y_pred = reg.predict(X_test)
r2_score = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test,y_pred)
print(f'The r2_score is: {r2_score:.2f}')
print(f'The mean squared error is: {mse:.2f}')
```


10. Use the cross validation function to assess the performance of a model on unseen data. Use the LinearRegression model.Make the number of folds 10.

```python
reg_2 = LinearRegression()
scores = cross_val_score(reg_2, X, y, cv=10, scoring ='r2')
print(f'The accuracy score is {scores.mean():.2f}')
```
