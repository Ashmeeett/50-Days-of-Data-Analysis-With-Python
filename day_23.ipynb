
1. Create a pandas DataFrame from the lists above. Write a code to check if there are any missing values in the DataFrame.

```python

import pandas as pd
import numpy as np


names = ["Kelly", np.nan, 'Jon', 'Ken', 'Tim', 'Pel']
grades = [30, 40, 30, 67, np.nan, 55]
age =[15, np.nan, 18, 17, np.nan, 16]

df = pd.DataFrame({'Names': names,'Grades':grades,'Age': age})
df
```

```python

# Checking for missing values
df.isnull().sum()
```


2. Your supervisor has asked that you show them how to fill the missing values, using the library scikit-learn. Write a code to fill missing values in the DataFrame above using this library. Use the most_frequent strategy for the "Names" column. Mean strategy for the "Grades" column. Median strategy for the "Age" column. Make a copy of the original DataFrame

```python
from sklearn.impute import SimpleImputer

df2 = df.copy()

# creating an instance of SimpleImputer with different strategies for each column
imputer_names = SimpleImputer( strategy='constant',fill_value="Paul",)
imputer_grades = SimpleImputer(strategy='mean')
imputer_age = SimpleImputer(strategy='median')

df2[['Names']] = imputer_names.fit_transform(df2[['Names']])
df2[['Grades']] = imputer_grades.fit_transform(df2[['Grades']])
df2[['Age']] = imputer_age.fit_transform(df2[['Age']])

df2
```


3. Using the original DataFrame (question 1) with missing values, identify and drop any columns that have more than 30% missing values. Save this to a new variable.

```python
# Check percentage of missing values in each column
missing_values = df.isnull().mean()

# Select columns that have more than 50% missing values
columns_to_drop = missing_values[missing_values > 0.30].index

# Drop columns from the dataframe
df_cleaned = df.drop(columns_to_drop, axis=1)

df_cleaned
```


4. Using pandas, check for any duplicate values in the "Names" columns of your DataFrame ( question 3).

```python
df_cleaned.duplicated("Names").sum()
```


5. Use the pandas groupby() method and agg() method to calculate the mean, minimum, and maximum of the "Grades" column, grouped by the "Names" column. Use the a copy of the DataFrame you saved in question 2.

```python
sch_data_agg = df2.groupby('Names').agg({'Grades': ['mean', 'min', 'max']})
sch_data_agg
```
