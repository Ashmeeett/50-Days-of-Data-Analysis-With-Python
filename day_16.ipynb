
1. Using pandas, create a pandas series with the list above. Write a code to find how many times each item appears in the list.

```python
import pandas as pd

list1 = ["wood", "red", "red", "white", "blue", "red"]

# Creating a pandas series
colors_series = pd.Series(list1)
colors_series
```

```python
colors_series.value_counts()
```


2. Using pandas, write a code to check if red, white, black is in the Series.

```python

colors = ["red", "white", "black"]

# Checking if the colors in the list are in series
is_in_series = colors_series.isin(colors)
is_in_series
```


3. Using pandas, write a code to return all unique values from the Series.

```python
unique_values = colors_series.unique()
unique_values
```


4. Using pandas update() method, write a code to update the Series you created in question 1. Add colors green and orange to the series. The color green will be at index 0 and orange will be at index 2.

```python
# Create a dictionary of updates
updates = {0: "green", 2: "orange"}

# Update the values in the Series
colors_series.update(updates)

# Print the updated Series
colors_series
```
