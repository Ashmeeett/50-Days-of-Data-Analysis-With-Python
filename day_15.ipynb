
1. You task is to import the CSV file using this NumPy’s genfromtxt() function.

```python
import numpy as np

arr = np.genfromtxt("Datasets_50_Days_Of_Data_Analysis/names_age_sex_data.csv", 
                    delimiter=",", dtype ='U7')
arr[:5]
```


a. Transpose the array.

```python

transposed_arr = arr.transpose()
transposed_arr
```


b. Using slicing, create three arrays: an array of all the names column, an array of the age column, and an array of the gender (sex) column from the array.

```python
names = transposed_arr[0:1,1:]
names
```

```python
age = transposed_arr[2:3,1:]
age
```

```python
gender = transposed_arr[1:2,1:]
gender
```


2. Using NumPy, write a code that returns all names of people age 56.

```python

aged_56 = names[age.astype(int) == 56]
print(aged_56)
```


3. How many people are age 44 or under?

```python

# Using Boolean slicing to access names
aged_44_under = names[age.astype(int) <= 44]

# Using the shape attribute to find people under 44
aged_44_under_num = np.shape(aged_44_under)
print("The number of people aged 44 and under is", 
      aged_44_under_num[0])
```


4. Write another code to return an array of all the males in the table. How many males are in the table?

```python

# Names of males in the array
males = names[gender =='Male']

# Number of males in the array
males_number_in_array = np.shape(males)
print(males)
print("The number of males is:", males_number_in_array[0])
```


5. Calculate the average age of all the females in the table.

```python
age = age.astype(int)

average_age = age[gender == 'Female'].mean()
print("The average females age is ", average_age)
```

6. Calculate the average age of the persons named Olivia and kate

```python
olivia_age = age[names == 'Olivia']
kate_age =  age[names == 'Kate']

# Create array and calculate the average age 
average = np.array([olivia_age, kate_age]).mean()
print(f'Average age of persons named Olivia and Kate is: {average:.2f}')
```
